```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# 1. Veriyi oku
df = pd.read_csv("all_data_ratio.csv", encoding='utf-8-sig')
```

```python
# 2. Sadece istediğin temel değişkenleri seç
temel_degiskenler = ['yıl', 'hastane', 'yatak', 'asistan', 'diger', 'ebe', 'eczaci', 'hemsire', 'nufus', 'pratisyen', 'uzman']
df_temel = df[temel_degiskenler]
```

```python
# 3. Korelasyon matrisi oluştur
corr = df_temel.corr()
```

```python
plt.figure(figsize=(12, 10))
sns.heatmap(
    corr, 
    annot=True,          # Sayıları yaz
    fmt=".2f",            # İki basamak
    cmap="coolwarm", 
    square=True, 
    linewidths=0.5, 
    cbar_kws={"shrink": .8}
)

plt.title("Temel Özellikler Arası Korelasyon Matrisi", fontsize=20, pad=20)
plt.xticks(rotation=45, ha='right', fontsize=10)
plt.yticks(rotation=0, fontsize=10)
plt.tight_layout()
plt.show()
```

```python
corr.to_csv("temel_ozellikler_korelasyon_matrisi.csv", encoding='utf-8-sig', float_format="%.2f")
```
