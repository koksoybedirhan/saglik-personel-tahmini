```python
import pandas as pd
```

```python
# 1. Veriyi oku
df = pd.read_csv("all_data.csv", encoding='utf-8-sig')
```

```python
# 2. İlgili veri tiplerini filtrele
df_filtered = df[df['veri_tipi'].isin(['personel', 'hastane_sayisi', 'yatak_sayisi', 'nufus'])]
```

```python
# 3. Pivot tabloya çevir
pivot_df = df_filtered.pivot_table(
    index=['il_adı', 'yıl'],
    columns=['kategori'],
    values='değer',
    aggfunc='sum'
).reset_index()
```

```python
# Kolon adlarını sadeleştir
pivot_df.columns.name = None
pivot_df = pivot_df.rename(columns={
    'uzman': 'uzman',
    'asistan': 'asistan',
    'pratisyen': 'pratisyen',
    'dis': 'disci',  # diş hekimi için dosya adında 'dis' olabilir
    'ebe': 'ebe',
    'hemsire': 'hemsire',
    'eczaci': 'eczaci',
    'diger': 'diger',
    'hastane': 'hastane',
    'yatak': 'yatak',
    'nufus': 'nufus'
})
```

```python
# 4. Eksik kolonları sıfırla (bazı şehirlerde bazı yıllarda kategori eksik olabilir)
for col in ['uzman', 'asistan', 'pratisyen', 'disci', 'ebe', 'hemsire', 'eczaci', 'diger', 'hastane', 'yatak', 'nufus']:
    if col not in pivot_df.columns:
        pivot_df[col] = 0
```

```python
# 5. Toplam sağlık personeli hesapla
pivot_df['toplam_saglikci'] = pivot_df[['uzman', 'asistan', 'pratisyen', 'disci', 'ebe', 'hemsire', 'eczaci', 'diger']].sum(axis=1)
```

```python
# 6. Doktor ve bakım personeli yoğunluğu
pivot_df['doktor_toplami'] = pivot_df[['uzman', 'asistan', 'pratisyen']].sum(axis=1)
pivot_df['bakim_personeli_toplami'] = pivot_df[['hemsire', 'ebe']].sum(axis=1)
```

```python
# 7. Nüfusa göre oranlar
pivot_df['uzman_nufus_orani'] = pivot_df['uzman'] / pivot_df['nufus']
pivot_df['asistan_nufus_orani'] = pivot_df['asistan'] / pivot_df['nufus']
pivot_df['pratisyen_nufus_orani'] = pivot_df['pratisyen'] / pivot_df['nufus']
pivot_df['disci_nufus_orani'] = pivot_df['disci'] / pivot_df['nufus']
pivot_df['ebe_nufus_orani'] = pivot_df['ebe'] / pivot_df['nufus']
pivot_df['hemsire_nufus_orani'] = pivot_df['hemsire'] / pivot_df['nufus']
pivot_df['eczaci_nufus_orani'] = pivot_df['eczaci'] / pivot_df['nufus']
pivot_df['diger_saglik_nufus_orani'] = pivot_df['diger'] / pivot_df['nufus']

pivot_df['yatak_hastane_orani'] = pivot_df['yatak'] / pivot_df['hastane']
pivot_df['hastane_nufus_orani'] = pivot_df['hastane'] / pivot_df['nufus']
pivot_df['yatak_nufus_orani'] = pivot_df['yatak'] / pivot_df['nufus']

pivot_df['uzman_yatak_orani'] = pivot_df['uzman'] / pivot_df['yatak']
pivot_df['asistan_yatak_orani'] = pivot_df['asistan'] / pivot_df['yatak']
pivot_df['pratisyen_yatak_orani'] = pivot_df['pratisyen'] / pivot_df['yatak']
pivot_df['disci_yatak_orani'] = pivot_df['disci'] / pivot_df['yatak']
pivot_df['ebe_yatak_orani'] = pivot_df['ebe'] / pivot_df['yatak']
pivot_df['hemsire_yatak_orani'] = pivot_df['hemsire'] / pivot_df['yatak']
pivot_df['eczaci_yatak_orani'] = pivot_df['eczaci'] / pivot_df['yatak']
pivot_df['diger_saglik_yatak_orani'] = pivot_df['diger'] / pivot_df['yatak']
```

```python
# 8. Yeni: Artış oranları hesaplama (nufus, uzman, yatak gibi)
pivot_df = pivot_df.sort_values(by=['il_adı', 'yıl'])  # sıralama önemli

pivot_df['nufus_artis_orani'] = pivot_df.groupby('il_adı')['nufus'].pct_change(fill_method=None)
pivot_df['uzman_artis_orani'] = pivot_df.groupby('il_adı')['uzman'].pct_change(fill_method=None)
pivot_df['asistan_artis_orani'] = pivot_df.groupby('il_adı')['asistan'].pct_change(fill_method=None)
pivot_df['pratisyen_artis_orani'] = pivot_df.groupby('il_adı')['pratisyen'].pct_change(fill_method=None)
pivot_df['disci_artis_orani'] = pivot_df.groupby('il_adı')['disci'].pct_change(fill_method=None)
pivot_df['ebe_artis_orani'] = pivot_df.groupby('il_adı')['ebe'].pct_change(fill_method=None)
pivot_df['hemsire_artis_orani'] = pivot_df.groupby('il_adı')['hemsire'].pct_change(fill_method=None)
pivot_df['eczaci_artis_orani'] = pivot_df.groupby('il_adı')['eczaci'].pct_change(fill_method=None)
pivot_df['diger_saglik_artis_orani'] = pivot_df.groupby('il_adı')['diger'].pct_change(fill_method=None)
pivot_df['yatak_artis_orani'] = pivot_df.groupby('il_adı')['yatak'].pct_change(fill_method=None)
```

```python
# 9. Yeni: Sağlıkçı yoğunlukları
pivot_df['bolge_basina_saglikci_yogunlugu'] = pivot_df['nufus'] / pivot_df['toplam_saglikci']
pivot_df['yatak_basina_saglikci_yogunlugu'] = pivot_df['toplam_saglikci'] / pivot_df['yatak']
```

```python
# 10. Yeni: Doktor ve Bakım personel yoğunlukları
pivot_df['doktor_nufus_orani'] = pivot_df['doktor_toplami'] / pivot_df['nufus']
pivot_df['bakim_personeli_nufus_orani'] = pivot_df['bakim_personeli_toplami'] / pivot_df['nufus']
```

```python
# Sadece 2002-2022 yıllarını bırak
pivot_df = pivot_df[(pivot_df['yıl'] >= 2002) & (pivot_df['yıl'] <= 2022)]
```

```python
# 11. Kaydet
pivot_df.to_csv("all_data_ratio.csv", index=False, encoding='utf-8-sig', float_format='%.10f')

print("✅ Gelişmiş özellik üretimi tamamlandı ve all_data_ratio.csv dosyasına kaydedildi!")
```
