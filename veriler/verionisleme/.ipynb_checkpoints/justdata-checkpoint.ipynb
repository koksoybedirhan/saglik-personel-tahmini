```python
import pandas as pd
```

```python
# 1. Ana veriyi oku
df = pd.read_csv("all_data_ratio.csv", encoding='utf-8-sig')
```

```python
# 2. Sadece temel değişkenleri seç
temel_degiskenler = ['yıl', 'nufus', 'hastane', 'yatak', 'asistan', 'diger', 'ebe', 'eczaci', 'hemsire', 'pratisyen', 'uzman']
df_temel = df[temel_degiskenler]
```

```python
# 3. Yeni CSV dosyasına kaydet
df_temel.to_csv("just_data.csv", index=False, encoding='utf-8-sig', float_format="%.8f")
```
