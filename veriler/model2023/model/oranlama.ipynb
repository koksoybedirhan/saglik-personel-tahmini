```python
import pandas as pd
```

```python
# 1. Veriyi oku
df = pd.read_csv("all_data2.csv", encoding='utf-8-sig')
```

```python
# 2. İlgili veri tiplerini filtrele
df_filtered = df[df['veri_tipi'].isin(['personel', 'hastane_sayisi', 'yatak_sayisi', 'nufus'])]
```

```python
# 3. Pivot tabloya çevir
pivot_df = df_filtered.pivot_table(
    index=['il_adı', 'yıl'],
    columns=['kategori'],
    values='değer',
    aggfunc='sum'
).reset_index()
```

```python
# Kolon adlarını sadeleştir
pivot_df.columns.name = None
pivot_df = pivot_df.rename(columns={
    'uzman': 'uzman',
    'asistan': 'asistan',
    'pratisyen': 'pratisyen',
    'dis': 'disci',  # diş hekimi için dosya adında 'dis' olabilir
    'ebe': 'ebe',
    'hemsire': 'hemsire',
    'eczaci': 'eczaci',
    'diger': 'diger',
    'hastane': 'hastane',
    'yatak': 'yatak',
    'nufus': 'nufus'
})
```

```python
# 4. Eksik kolonları sıfırla (bazı şehirlerde bazı yıllarda kategori eksik olabilir)
for col in ['uzman', 'asistan', 'pratisyen', 'disci', 'ebe', 'hemsire', 'eczaci', 'diger', 'hastane', 'yatak', 'nufus']:
    if col not in pivot_df.columns:
        pivot_df[col] = 0
```

```python
# 5. Toplam sağlık personeli hesapla
pivot_df['toplam_saglikci'] = pivot_df[['uzman', 'asistan', 'pratisyen', 'disci', 'ebe', 'hemsire', 'eczaci', 'diger']].sum(axis=1)
```

```python
# 6. Doktor ve bakım personeli yoğunluğu
pivot_df['doktor_toplami'] = pivot_df[['uzman', 'asistan', 'pratisyen']].sum(axis=1)
pivot_df['bakim_personeli_toplami'] = pivot_df[['hemsire', 'ebe']].sum(axis=1)
```

```python
# Sadece 2002-2022 yıllarını bırak
pivot_df = pivot_df[(pivot_df['yıl'] >= 2002) & (pivot_df['yıl'] <= 2024)]
```

```python
# 11. Kaydet
pivot_df.to_csv("all_data_ratio.csv", index=False, encoding='utf-8-sig', float_format='%.8f')

print("✅ Gelişmiş özellik üretimi tamamlandı ve all_data_ratio.csv dosyasına kaydedildi!")
```
