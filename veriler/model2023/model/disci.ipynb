```python
import pandas as pd

# CSV dosyasını oku
df = pd.read_csv("all_data.csv")

# 'kategori' sütununda 'disci' olanları 'dis' olarak değiştir
df["kategori"] = df["kategori"].replace("disci", "dis")

# Değişiklikleri kontrol et (opsiyonel)
print(df["kategori"].value_counts())

# Gerekirse yeni dosya olarak kaydet
df.to_csv("all_data2.csv", index=False)
```
