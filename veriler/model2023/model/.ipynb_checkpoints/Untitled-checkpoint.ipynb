```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor

# Türkiye'nin 7 coğrafi bölgesi
bolge_haritasi = {
    'Marmara': ["İstanbul", "Edirne", "Kırklareli", "Tekirdağ", "Kocaeli", "Sakarya", "Yalova", "Bilecik", "Balıkesir", "Çanakkale", "Bursa"],
    'Ege': ["İzmir", "Aydın", "Muğla", "Manisa", "Afyonkarahisar", "Uşak", "Kütahya", "Denizli"],
    'Akdeniz': ["Antalya", "Mersin", "Adana", "Hatay", "Kahramanmaraş", "Osmaniye", "Isparta", "Burdur"],
    'İç Anadolu': ["Ankara", "Konya", "Kayseri", "Eskişehir", "Nevşehir", "Aksaray", "Kırıkkale", "Niğde", "Karaman", "Sivas", "Yozgat", "Kırşehir", "Çankırı"],
    'Karadeniz': ["Samsun", "Trabzon", "Ordu", "Giresun", "Rize", "Artvin", "Tokat", "Amasya", "Bartın", "Zonguldak", "Karabük", "Sinop", "Çorum", "Bayburt", "Gümüşhane", "Kastamonu", "Bolu"],
    'Doğu Anadolu': ["Erzurum", "Van", "Malatya", "Elazığ", "Kars", "Ağrı", "Bitlis", "Iğdır", "Tunceli", "Ardahan", "Erzincan", "Bingöl", "Muş", "Hakkari"],
    'Güneydoğu Anadolu': ["Diyarbakır", "Gaziantep", "Şanlıurfa", "Mardin", "Batman", "Siirt", "Şırnak", "Adıyaman", "Kilis"]
}

def bolge_bul(il):
    for bolge, iller in bolge_haritasi.items():
        if il in iller:
            return bolge
    return "Bilinmiyor"

# Özel feature setleri
feature_map = {
    "uzman": ["yıl", "nufus", "hastane", "yatak", "asistan", "pratisyen", "hemsire", "ebe", "diger"],
    "pratisyen": ["yıl", "nufus", "hastane", "yatak", "asistan", "uzman", "hemsire", "ebe", "diger"],
    "asistan": ["yıl", "nufus", "hastane", "yatak", "uzman", "pratisyen", "hemsire", "ebe", "diger"],
    "ebe": ["yıl", "nufus", "hastane", "yatak", "asistan", "uzman", "hemsire", "pratisyen", "diger"],
    "diger": ["yıl", "nufus", "hastane", "yatak", "asistan", "pratisyen", "hemsire", "ebe", "uzman"],
    "hemsire": ["yıl", "nufus", "hastane", "yatak", "asistan", "uzman", "pratisyen", "ebe", "diger"],
    "yatak": ["yıl", "nufus", "hastane", "hemsire", "asistan", "uzman", "pratisyen", "ebe", "diger"],
    "hastane": ["yıl", "nufus", "yatak", "hemsire", "asistan", "uzman", "pratisyen", "ebe", "diger"]
}

# Veriyi oku ve düzenle
df = pd.read_csv("all_data.csv")
df["kategori"] = df["kategori"].replace("disci", "dis")  # disci → dis
df = df[~df["kategori"].isin(["dis", "eczaci"])]         # Dişçi ve eczacı çıkarıldı

# Pivot işlemleri
df_nufus = df[df["veri_tipi"] == "nufus"][["il_adı", "yıl", "değer"]].rename(columns={"değer": "nufus"})
df_other = df[df["veri_tipi"] != "nufus"]
df_pivot = df_other.pivot_table(index=["il_adı", "yıl"], columns=["veri_tipi", "kategori"], values="değer", aggfunc="sum")
df_pivot.columns = ['_'.join(col).strip() for col in df_pivot.columns.values]
df_pivot = df_pivot.reset_index()
df_merged = pd.merge(df_pivot, df_nufus, on=["il_adı", "yıl"], how="left").dropna()
df_merged["bolge"] = df_merged["il_adı"].apply(bolge_bul)

# Hedef değişken listesi (dişçi, eczacı hariç)
hedefler = list(feature_map.keys())
tahminler = []

for il in df_merged["il_adı"].unique():
    df_il = df_merged[df_merged["il_adı"] == il]
    if df_il[df_il["yıl"] <= 2021].shape[0] < 5:
        continue

    for hedef in hedefler:
        if hedef not in df_il.columns:
            continue

        features = feature_map[hedef]
        features = [f"personel_{f}" if f in hedefler else f for f in features]
        hedef_col = f"personel_{hedef}" if hedef in hedefler else hedef

        if hedef_col not in df_il.columns or any(f not in df_il.columns for f in features):
            continue

        df_target = df_il[features + [hedef_col]].dropna()
        if df_target.empty or df_target[df_il["yıl"] <= 2021].shape[0] == 0:
            continue

        train = df_target[df_il["yıl"] <= 2021]
        test = df_target[df_il["yıl"] == 2023]
        if test.empty:
            continue

        X_train, y_train = train[features], train[hedef_col]
        X_test, y_test = test[features], test[hedef_col]

        # Modeller
        lr = LinearRegression().fit(X_train, y_train)
        rf = RandomForestRegressor(n_estimators=100, random_state=42).fit(X_train, y_train)

        y_pred_lr = lr.predict(X_test)
        y_pred_rf = rf.predict(X_test)

        tahminler.append({
            "il_adı": il,
            "bolge": bolge_bul(il),
            "hedef": hedef,
            "gercek_2023": y_test.values[0],
            "tahmin_2023_lr": y_pred_lr[0],
            "tahmin_2023_rf": y_pred_rf[0]
        })

# Bölgesel toplama
df_tahmin = pd.DataFrame(tahminler)
bolgesel = df_tahmin.groupby(["bolge", "hedef"])[["gercek_2023", "tahmin_2023_lr", "tahmin_2023_rf"]].sum().reset_index()

# Sonuçları kaydet
bolgesel.to_csv("bolgesel_2023_lr_rf_tahmin.csv", index=False)
```
