```python
import pandas as pd
```

```python
# 1. Veriyi oku
df = pd.read_csv("all_data.csv", encoding='utf-8-sig')
```

```python
# 2. Eksik veri kontrolü
print("📋 Eksik veri sayısı (kolon bazlı):")
print(df.isnull().sum())
```

```python
# 3. Veri tipi kontrolü
print("\n📋 Veri tipleri:")
print(df.dtypes)
```

```python
# 4. Genel özet istatistikler (sadece 'değer' kolonuna bakalım)
print("\n📋 Genel İstatistikler:")
print(df['değer'].describe())
```

```python
# 5. Hangi veri tiplerinde eksik veri var? (detaylı grup analizi)
eksik_veri_gruplu = df.groupby(['veri_tipi', 'kategori', 'kurum']).apply(lambda g: g['değer'].isnull().sum())
print("\n📋 Veri tipi bazlı eksik veri analizi:")
print(eksik_veri_gruplu[eksik_veri_gruplu > 0])
```

```python
# 6. Şehir-yıl bazlı eksik kayıtlar
eksik_sehir_yil = df[df['değer'].isnull()]
print("\n📋 Eksik değer bulunan şehir-yıl kayıtları:")
print(eksik_sehir_yil[['il_adı', 'yıl', 'veri_tipi', 'kategori', 'kurum']])
```

```python
# 7. Gözle hızlıca anormallik yakalamak için büyük/küçük değerler
print("\n📋 En yüksek 10 değer:")
print(df.sort_values(by='değer', ascending=False).head(10))
```

```python
print("\n📋 En düşük 10 değer:")
print(df.sort_values(by='değer', ascending=True).head(10))
```
