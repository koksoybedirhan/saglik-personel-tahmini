```python
import pandas as pd
```

```python
# 1. Dosyaları yükle
ihtiyaclar_path = 'ihtiyaclar.csv'          # Yeni mezunların şehir dağılım yüzdeleri
veriler_path = 'tahminsonuclari.csv'         # 2002-2024 arası hastane/yatak/personel verileri

ihtiyaclar = pd.read_csv(ihtiyaclar_path)
veriler = pd.read_csv(veriler_path)
```

```python
# 2. Kullanılacak personel türleri
personel_turleri = [
    "uzman", "asistan", "pratisyen", "diger", "hemsire", "ebe", "disci", "eczaci"
]
```

```python
# 3. 2024 verilerini al
veriler_2024 = veriler[veriler['yıl'] == 2024].copy()
```

```python
# 4. Türkiye genelinde 2023-2024 artışlarını hesapla
artis_df = veriler[veriler['yıl'] == 2024].copy()
for personel in personel_turleri:
    artis_df[personel] = veriler[veriler['yıl'] == 2024][personel].values - veriler[veriler['yıl'] == 2023][personel].values

turkiye_artislar = artis_df[personel_turleri].sum()
```

```python
# 5. 'Toplam' satırını kaldır
ihtiyaclar = ihtiyaclar[ihtiyaclar['il_adı'] != 'Toplam']
```

```python
# 6. Şehirlere göre yeni mezun dağılımı
eklenen_personel = ihtiyaclar.copy()
for personel in personel_turleri:
    eklenen_personel[personel] = (ihtiyaclar[personel] / 100) * turkiye_artislar[personel]
```

```python
# 7. 2025 verisini oluştur: mevcut 2024 verisine yeni mezunları ekle
veriler_2025 = veriler_2024.copy()
for personel in personel_turleri:
    veriler_2025[personel] = veriler_2024[personel].values + eklenen_personel[personel].values
```

```python
# 8. 2025 yılı için yeni personel/yatak ve personel/hastane oranlarını hesapla
for personel in personel_turleri:
    veriler_2025[f"{personel}_per_yatak"] = veriler_2025[personel] / veriler_2025["yatak"]
    veriler_2025[f"{personel}_per_hastane"] = veriler_2025[personel] / veriler_2025["hastane"]
```

```python
# 9. Sonuçları kaydet
veriler_2025.to_csv('2025_personel_oranlari.csv', index=False, float_format='%.8f')

print("✅ İşlem tamamlandı. Dosya '2025_personel_oranlari.csv' olarak kaydedildi.")
```
