```python
import pandas as pd
from sklearn.linear_model import LinearRegression

# 1. Dosyayı oku
file_path = '2025verileri.csv'
df = pd.read_csv(file_path)

# 2. İlgili personel türleri
personel_turleri = [
    "uzman", "asistan", "pratisyen", "diger", "hemsire", "ebe", "disci", "eczaci"
]

# 3. 2023 ve 2024 verilerini al
df_2023 = df[df['yıl'] == 2023]
df_2024 = df[df['yıl'] == 2024]

# 4. Şehir bazlı 2025 tahminleri çıkar
sehirler = df['il_adı'].unique()
tahmin_2025_listesi = []

for sehir in sehirler:
    sehir_veri = df[df['il_adı'] == sehir]
    
    tahmin_satiri = {'il_adı': sehir}
    
    for personel in personel_turleri:
        X = sehir_veri[['yıl']]
        y = sehir_veri[personel]
        
        model = LinearRegression()
        model.fit(X, y)
        
        tahmin_degeri = model.predict(pd.DataFrame({'yıl': [2025]}))[0]
        tahmin_satiri[personel] = max(tahmin_degeri, 0)
    
    tahmin_2025_listesi.append(tahmin_satiri)

df_2025 = pd.DataFrame(tahmin_2025_listesi)

# 5. 2023-2024 artışları
artis_2024 = df_2024.copy()
for personel in personel_turleri:
    artis_2024[personel] = df_2024[personel].values - df_2023[personel].values

artis_2024 = artis_2024[['il_adı'] + personel_turleri]

# 6. 2024-2025 artışları
artis_2025 = df_2025.copy()
for personel in personel_turleri:
    artis_2025[personel] = df_2025[personel].values - df_2024[personel].values

artis_2025 = artis_2025[['il_adı'] + personel_turleri]

# 7. Türkiye genelinde toplam artışları bul
turkiye_artislar_2024 = artis_2024[personel_turleri].sum()
turkiye_artislar_2025 = artis_2025[personel_turleri].sum()

# 8. Şehirlerin yüzdelerini hesapla (önce 2024 için)
ihtiyac_yuzdeleri_2024 = artis_2024.copy()

for personel in personel_turleri:
    ihtiyac_yuzdeleri_2024[personel] = (artis_2024[personel] / turkiye_artislar_2024[personel]) * 100

ihtiyac_yuzdeleri_2024 = ihtiyac_yuzdeleri_2024.round(8)
ihtiyac_yuzdeleri_2024['tahmin_yili'] = 2024

# 9. Şehirlerin yüzdelerini hesapla (şimdi 2025 için)
ihtiyac_yuzdeleri_2025 = artis_2025.copy()

for personel in personel_turleri:
    ihtiyac_yuzdeleri_2025[personel] = (artis_2025[personel] / turkiye_artislar_2025[personel]) * 100

ihtiyac_yuzdeleri_2025 = ihtiyac_yuzdeleri_2025.round(8)
ihtiyac_yuzdeleri_2025['tahmin_yili'] = 2025

# 10. Verileri birleştir
son_df = pd.concat([ihtiyac_yuzdeleri_2024, ihtiyac_yuzdeleri_2025], ignore_index=True)

# 11. Kolonları sırala
son_df = son_df[['tahmin_yili', 'il_adı'] + personel_turleri]

# 12. Şehir ve yıl sıralamasını yap
son_df = son_df.sort_values(by=['il_adı', 'tahmin_yili']).reset_index(drop=True)

# 13. CSV'ye kaydet
son_df.to_csv(
    'son.csv',
    index=False,
    float_format='%.8f'
)

# 14. Kontrol: 2024 ve 2025 için toplamlar ayrı yaz
print("2024 yılı toplam yüzdeleri:")
print(son_df[son_df['tahmin_yili'] == 2024][personel_turleri].sum())

print("\n2025 yılı toplam yüzdeleri:")
print(son_df[son_df['tahmin_yili'] == 2025][personel_turleri].sum())

print("\n✅ 2024 ve 2025 verileri ayrı normalize edildi ve düzgün sıralandı!")
```
