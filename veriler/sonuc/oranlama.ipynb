```python
import pandas as pd

# Dosyayı oku
file_path = 'tahminsonuclari.csv'
df = pd.read_csv(file_path)
```

```python
# Hangi personel türlerine oran çıkaracağımızı listeleyelim
personel_turleri = [
    "uzman", "asistan", "pratisyen", "diger", "hemsire", "ebe", "disci", "eczaci"
]
```

```python
# Her personel için yatak ve hastane oranı hesapla
for personel in personel_turleri:
    df[f"{personel}_per_yatak"] = df[personel] / df["yatak"]
    df[f"{personel}_per_hastane"] = df[personel] / df["hastane"]
```

```python
# Sonucu kontrol için yazdır
print(df.head())
```

```python
# İstersen yeni CSV dosyası olarak da kaydedebilirsin:
df.to_csv('tahmin-oran.csv', index=False)
```
