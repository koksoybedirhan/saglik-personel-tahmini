```python
import pandas as pd
```

```python
df = pd.read_csv("nufüs2324-temizleme.csv", sep=";", header=None)
```

```python
df.columns = df.iloc[0]
df = df.drop(0).reset_index(drop=True)
```

```python
df.rename(columns={df.columns[0]: "il_adı"}, inplace=True)
```

```python
for col in df.columns:
    df[col] = df[col].astype(str).str.strip().str.rstrip(",")
```

```python
for col in df.columns[1:]:
    df[col] = (
        df[col]
        .astype(str)
        .str.replace(" ", "", regex=False)
        .str.replace(".", "", regex=False)
        .str.replace("nan", "", regex=False)
    )
    df[col] = pd.to_numeric(df[col], errors="coerce").astype("Int64")  # Nullable integer
```

```python
df.to_csv("nufus2324_temizlenmis_utf8_clean.csv", index=False, encoding="utf-8-sig")
```
