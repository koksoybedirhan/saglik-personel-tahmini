```python
import pandas as pd
```

```python
df_nufus = pd.read_csv("nufustemizleme.csv", sep=";", header=None)
```

```python
df_nufus.columns = df_nufus.iloc[0]
df_nufus = df_nufus.drop(0).reset_index(drop=True)
```

```python
df_nufus.rename(columns={df_nufus.columns[0]: "il_adı"}, inplace=True)
```

```python
for col in df_nufus.columns[1:]:
    df_nufus[col] = (
        df_nufus[col]
        .astype(str)
        .str.replace(" ", "", regex=False)
        .str.replace(".", "", regex=False)
        .astype(int)
    )
```

```python
df_nufus.to_csv("nufus_temizlenmis_utf8.csv", index=False, encoding="utf-8-sig")
```
