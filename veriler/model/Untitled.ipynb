```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
```

```python
# 1. Ana veriyi oku
df = pd.read_csv("all_data_ratio.csv", encoding='utf-8-sig')

# 2. 2023-2024 nüfus verisini oku
nufus_2324 = pd.read_csv("nufus2324.csv", encoding='utf-8-sig')
# nufüs2324 dosyasının adlarının 'il_adı', 'yıl', 'nufus' olduğunu varsayıyorum
# Eğer değilse sütun adlarını uydururuz.
```

```python
# 3. Tüm şehirler için tahminleri toplayacağımız liste
tahmin_sonuclari = []
```

```python
# 4. Şehirleri sırayla işle
sehirler = df['il_adı'].unique()

for sehir in sehirler:
    # Şehrin 2002-2022 verisini al
    df_sehir = df[df['il_adı'] == sehir]
    
    # Eğer veri eksikse atla
    if df_sehir.shape[0] < 5:
        continue

    # X ve y hazırla
    X = df_sehir[['yıl', 'nufus']]
    y = df_sehir['uzman']

    # Modelleri eğit
    lr_model = LinearRegression()
    rf_model = RandomForestRegressor(random_state=42, n_estimators=100)
    
    lr_model.fit(X, y)
    rf_model.fit(X, y)

    # Şehrin 2023 ve 2024 nüfus verisini al
    try:
        nufus_sehir = nufus_2324[nufus_2324['il_adı'] == sehir]
        X_future = nufus_sehir[['yıl', 'nufus']]
        
        # Linear Regression ve Random Forest tahminleri yap
        lr_preds = lr_model.predict(X_future)
        rf_preds = rf_model.predict(X_future)

        # Sonuçları listeye ekle
        for idx, yil in enumerate(X_future['yıl']):
            tahmin_sonuclari.append({
                'il_adı': sehir,
                'yıl': yil,
                'nufus': X_future.iloc[idx]['nufus'],
                'Linear_Regression_Tahmini': lr_preds[idx],
                'Random_Forest_Tahmini': rf_preds[idx]
            })

    except Exception as e:
        print(f"{sehir} için 2023-2024 nüfus verisi bulunamadı. Atlandı.")
```

```python
# 5. Tahmin sonuçlarını DataFrame yap
df_tahmin = pd.DataFrame(tahmin_sonuclari)
```

```python
# 6. Excel dosyasına kaydet
df_tahmin.to_excel("tum_sehirler_uzman_tahminleri_2023_2024.xlsx", index=False)

print("✅ Tahminler başarıyla 'tum_sehirler_uzman_tahminleri_2023_2024.xlsx' dosyasına kaydedildi!")
```
