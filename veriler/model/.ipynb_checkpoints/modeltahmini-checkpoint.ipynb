```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
```

```python
# 1. Güncellenmiş ana veri dosyasını oku
df = pd.read_csv("all_data_ratio_2023_2024_eklenmis.csv", encoding='utf-8-sig')
```

```python
# 2. Tüm şehirler için tahminleri toplayacağımız liste
tahmin_sonuclari = []
```

```python
sehirler = df['il_adı'].unique()

for sehir in sehirler:
    # Şehrin 2002-2022 verisini seç
    df_sehir = df[(df['il_adı'] == sehir) & (df['yıl'] <= 2022)]

    # Eğer çok az veri varsa atla
    if df_sehir.shape[0] < 5:
        continue

    # X = sadece yıl ve nufus
    X = df_sehir[['yıl', 'nufus']]
    y = df_sehir['uzman']

    # Model eğit
    lr_model = LinearRegression()

    lr_model.fit(X, y)

    # 2023 ve 2024 verisini çek
    df_future = df[(df['il_adı'] == sehir) & (df['yıl'] >= 2023)]

    if df_future.shape[0] == 2:
        X_future = df_future[['yıl', 'nufus']]  # SADECE yıl ve nüfus ile tahmin yapıyoruz!

        # Tahminler
        lr_preds = lr_model.predict(X_future)

        # Tahminleri kaydet
        for idx, yil in enumerate(X_future['yıl']):
            tahmin_sonuclari.append({
                'il_adı': sehir,
                'yıl': yil,
                'nufus': X_future.iloc[idx]['nufus'],
                'Linear_Regression_Tahmini': lr_preds[idx],
            })
```

```python
# 4. Tahmin sonuçlarını DataFrame yap
df_tahmin = pd.DataFrame(tahmin_sonuclari)
```

```python
# 5. Excel dosyasına kaydet
df_tahmin.to_csv("uzmantahmin.csv", index=False, encoding='utf-8-sig')

print("✅ Tahminler başarıyla 'tum_sehirler_uzman_tahminleri_2023_2024_guncel.xlsx' dosyasına kaydedildi!")
```
