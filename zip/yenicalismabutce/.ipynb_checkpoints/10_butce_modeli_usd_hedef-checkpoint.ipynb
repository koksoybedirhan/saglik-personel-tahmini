```python
# =====================================================================
# 10 - 1999-2024 TÜRKİYE SAĞLIK BÜTÇESİ TAHMİNİ
#
# GÜNCEL YAKLAŞIM:
#   * Hedef:    butce_USD (Toplam Sağlık Harcaması, Milyon USD)
#   * Sebep:    TL bazlı hedef, yüksek enflasyon ve kur etkisi nedeniyle
#               recursive prediction sırasında exponential şişme üretebilir.
#   * 2025-2030 için model önce USD bazlı reel bütçeyi tahmin eder.
#   * USD/TRY değeri MODELLE TAHMİN EDİLMEZ.
#   * 2025-2030 USD/TRY gelecek kur verisi doğrudan
#     usdtry_aylik_tahmin_2026_2030.csv dosyasından alınır.
#   * TL bütçe sonradan türetilir:
#       butce_TL_turetilen = butce_USD_tahmin × usdtry_hazir_veri
#
# Girdiler:
#   - 99-24saglikbutcesi_temiz.xlsx
#   - usd-try.csv
#   - usdtry_aylik_tahmin_2026_2030.csv
# =====================================================================

# %% [1] KÜTÜPHANELER ================================================
import warnings, time, os
warnings.filterwarnings("ignore")
import numpy as np
import pandas as pd

from sklearn.linear_model    import LinearRegression, Ridge, ElasticNet
from sklearn.ensemble        import (RandomForestRegressor, ExtraTreesRegressor,
                                     GradientBoostingRegressor, BaggingRegressor,
                                     VotingRegressor, StackingRegressor,
                                     HistGradientBoostingRegressor)
from sklearn.svm             import SVR
from sklearn.neighbors       import KNeighborsRegressor
from sklearn.preprocessing   import StandardScaler
from sklearn.metrics         import (mean_absolute_error, mean_squared_error, r2_score)
from sklearn.model_selection import GridSearchCV

from xgboost import XGBRegressor
try:
    from catboost import CatBoostRegressor
    HAVE_CAT = True
except ImportError:
    HAVE_CAT = False

from statsmodels.tsa.holtwinters import ExponentialSmoothing
from statsmodels.tsa.arima.model  import ARIMA
try:
    from prophet import Prophet
    HAVE_PROPHET = True
except ImportError:
    HAVE_PROPHET = False


```

```python
# %% [2] BÜTÇE VERİSİ ================================================

import os
import pandas as pd

BUTCE_KAYNAKLAR = [
    "99-24saglikbutcesi_temiz.xlsx",
    "99-24saglikbutcesi_temiz.csv",
    "99-24saglikbutcesi.xlsx",
    "99-24saglikbutcesi.csv",
]

butce_path = next((p for p in BUTCE_KAYNAKLAR if os.path.exists(p)), None)

if butce_path is None:
    raise FileNotFoundError("Bütçe dosyası bulunamadı.")

if butce_path.endswith((".xls", ".xlsx")):
    raw = pd.read_excel(butce_path, header=2)
else:
    raw = pd.read_csv(butce_path, header=2)

raw.columns = [str(c).strip() for c in raw.columns]

print(f"[OK] Bütçe okundu: {butce_path}")
print("Sütunlar:", raw.columns.tolist())

butce = raw.rename(columns={
    "Yıl": "yıl",
    "Toplam sağlık harcaması (Milyon TL)": "butce_TL",
    "Kişi başına sağlık harcaması (TL)": "kisi_basi_TL",
    "Genel devlet sağlık harcaması oranı (%)": "devlet_orani",
    "Özel sektör sağlık harcaması oranı (%)": "ozel_orani"
}).copy()

butce = butce[["yıl", "butce_TL", "kisi_basi_TL", "devlet_orani", "ozel_orani"]]

for col in butce.columns:
    butce[col] = pd.to_numeric(
        butce[col].astype(str)
        .str.replace(",", ".", regex=False)
        .str.replace(" ", "", regex=False),
        errors="coerce"
    )

butce = butce.dropna(subset=["yıl", "butce_TL"]).copy()
butce["yıl"] = butce["yıl"].astype(int)

butce = butce[
    (butce["yıl"] >= 1999) &
    (butce["yıl"] <= 2024)
].sort_values("yıl").reset_index(drop=True)

print("\nBütçe yıllık veri:")
display(butce)
```

```python
# %% [3] USD/TRY KURU — yıllık ortalama =============================
KUR_PATHS = ["usd-try.csv", "usd-try-8a0a40c9.csv"]
kur_path = next((p for p in KUR_PATHS if os.path.exists(p)), None)
if kur_path is None:
    raise FileNotFoundError("usd-try.csv bulunamadı.")

kur = pd.read_csv(kur_path)
kur["tarih"] = pd.to_datetime(kur["Tarih"], format="%d.%m.%Y", errors="coerce")
kur["kur"] = pd.to_numeric(
    kur["Şimdi"].astype(str).str.replace(",", ".").str.replace(" ", ""),
    errors="coerce")
kur["yıl"] = kur["tarih"].dt.year
kur_yillik = kur.groupby("yıl")["kur"].mean().reset_index()
kur_yillik.columns = ["yıl", "usdtry"]
print(f"\n[OK] Kur okundu: {kur_path}, yıl aralığı: "
      f"{kur_yillik['yıl'].min()}-{kur_yillik['yıl'].max()}")


```

```python
# %% [4] BİRLEŞTİR + TÜREV ÖZELLİKLER ================================
veri = butce.merge(kur_yillik, on="yıl", how="left")

eksik = veri.loc[veri["usdtry"].isna(), "yıl"].tolist()
if eksik:
    raise ValueError(f"USD/TRY kuru eksik yıllar: {eksik}")

# USD bazlı reel bütçe (gözlemlenmiş — gelecek için tahmin değil)
veri["butce_USD"]      = veri["butce_TL"] / veri["usdtry"]
veri["kisi_basi_USD"]  = veri["kisi_basi_TL"] / veri["usdtry"]

# Lag özellikleri (1-3 yıl geçmiş + yıllık değişim oranları)
veri = veri.sort_values("yıl").reset_index(drop=True)
for col in ["butce_TL", "kisi_basi_TL", "devlet_orani", "ozel_orani",
            "butce_USD", "kisi_basi_USD", "usdtry"]:
    veri[f"{col}_lag1"] = veri[col].shift(1)
    veri[f"{col}_lag2"] = veri[col].shift(2)
    veri[f"{col}_yoy"]  = veri[col].pct_change()

print("\nLag eklenmiş veri:")
print(veri.tail(6).to_string(index=False))
veri.to_csv("butce_veri_hazir.csv", index=False, encoding="utf-8-sig")


```

```python
# %% [5] HEDEF + FEATURE ============================================
# ÖNEMLİ DEĞİŞİKLİK:
#   Eski hedef: butce_TL
#   Yeni hedef: butce_USD
#
# Amaç:
#   TL hedefte recursive tahmin exponential şişme üretebiliyor.
#   USD hedef daha stabil olduğu için önce reel bütçe tahmin edilir,
#   sonra hazır USD/TRY kuru ile TL bütçe türetilir.

HEDEF = "butce_USD"   # Toplam Sağlık Harcaması (Milyon USD) — ana model hedefi

# Features:
#   TL hedef lag'leri çıkarıldı. Çünkü patlamanın ana kaynağı TL lag-recursive yapıydı.
#   USD bazlı lag/yoy değerleri ana sinyal olarak kullanılıyor.
FEATURES = [
    "yıl",

    # USD bazlı reel sağlık harcaması geçmişi
    "butce_USD_lag1",
    "butce_USD_lag2",
    "butce_USD_yoy",

    # Kişi başına sağlık harcaması — USD bazlı daha stabil sinyal
    "kisi_basi_USD_lag1",
    "kisi_basi_USD_lag2",
    "kisi_basi_USD_yoy",

    # Harcama kompozisyonu
    "devlet_orani_lag1",
    "ozel_orani_lag1",

    # Kur bilgisi feature olarak kalır; fakat gelecek kur modelle tahmin edilmez
    "usdtry_lag1",
    "usdtry_yoy",
]

print(f"Hedef:    {HEDEF}")
print(f"Features ({len(FEATURES)}): {FEATURES}")


```

```python
# %% [6] K-FOLD (ROLLING) ===========================================
# 26 yıl veri (1999-2024) → 5 fold, her test 3 yıl
folds = [
    (range(1999, 2017), range(2017, 2020)),
    (range(1999, 2018), range(2018, 2021)),
    (range(1999, 2019), range(2019, 2022)),
    (range(1999, 2020), range(2020, 2023)),
    (range(1999, 2022), range(2022, 2025)),    # Fold 5 — ANA FOLD
]
print(f"Toplam {len(folds)} fold")


```

```python
# %% [7] MODEL REGISTRY (14 model) ==================================
INNER_CV = 2
SCORING  = "neg_mean_absolute_error"
models_registry = {
    "LinearRegression": lambda: GridSearchCV(LinearRegression(),
        {"fit_intercept": [True, False]}, cv=INNER_CV, scoring=SCORING),
    "Ridge": lambda: GridSearchCV(Ridge(random_state=42),
        {"alpha": [0.1, 1.0, 10.0]}, cv=INNER_CV, scoring=SCORING),
    "ElasticNet": lambda: GridSearchCV(ElasticNet(random_state=42, max_iter=5000),
        {"alpha": [0.1, 1.0], "l1_ratio": [0.3, 0.5, 0.7]}, cv=INNER_CV, scoring=SCORING),
    "RandomForest": lambda: GridSearchCV(RandomForestRegressor(random_state=42, n_jobs=-1),
        {"n_estimators": [100, 200], "max_depth": [None, 5]}, cv=INNER_CV, scoring=SCORING),
    "ExtraTrees": lambda: GridSearchCV(ExtraTreesRegressor(random_state=42, n_jobs=-1),
        {"n_estimators": [100, 200], "max_depth": [None, 5]}, cv=INNER_CV, scoring=SCORING),
    "GradientBoosting": lambda: GridSearchCV(GradientBoostingRegressor(random_state=42),
        {"n_estimators": [100, 200], "learning_rate": [0.05, 0.1]}, cv=INNER_CV, scoring=SCORING),
    "HistGradientBoosting": lambda: GridSearchCV(HistGradientBoostingRegressor(random_state=42),
        {"max_iter": [100, 200], "learning_rate": [0.05, 0.1]}, cv=INNER_CV, scoring=SCORING),
    "XGBoost": lambda: GridSearchCV(
        XGBRegressor(random_state=42, verbosity=0, objective="reg:squarederror"),
        {"n_estimators": [200, 300], "learning_rate": [0.05, 0.1]}, cv=INNER_CV, scoring=SCORING),
    "SVR": lambda: GridSearchCV(SVR(),
        {"C": [1, 10], "epsilon": [0.05, 0.1]}, cv=INNER_CV, scoring=SCORING),
    "KNN": lambda: GridSearchCV(KNeighborsRegressor(),
        {"n_neighbors": [3, 5], "weights": ["uniform", "distance"]}, cv=INNER_CV, scoring=SCORING),
    "Bagging": lambda: GridSearchCV(BaggingRegressor(random_state=42, n_jobs=-1),
        {"n_estimators": [50, 100]}, cv=INNER_CV, scoring=SCORING),
    "Voting": lambda: VotingRegressor([
        ("rf", RandomForestRegressor(n_estimators=200, random_state=42, n_jobs=-1)),
        ("xgb", XGBRegressor(n_estimators=200, random_state=42, verbosity=0,
                             objective="reg:squarederror")),
        ("gb", GradientBoostingRegressor(n_estimators=200, random_state=42))]),
    "Stacking": lambda: StackingRegressor(
        estimators=[
            ("rf", RandomForestRegressor(n_estimators=200, random_state=42, n_jobs=-1)),
            ("xgb", XGBRegressor(n_estimators=200, random_state=42, verbosity=0,
                                 objective="reg:squarederror"))],
        final_estimator=LinearRegression()),
}
if HAVE_CAT:
    models_registry["CatBoost"] = lambda: GridSearchCV(
        CatBoostRegressor(verbose=0, random_seed=42),
        {"iterations": [200, 300], "depth": [4, 6]}, cv=INNER_CV, scoring=SCORING)

UNI_MODELS = ["HoltWinters", "ARIMA"]
if HAVE_PROPHET: UNI_MODELS.append("Prophet")
print(f"Toplam {len(models_registry) + len(UNI_MODELS)} model")


```

```python
# %% [8] METRİK =====================================================
def metrikler(y_true, y_pred):
    y_true = np.asarray(y_true, dtype=float)
    y_pred = np.asarray(y_pred, dtype=float)
    mae  = mean_absolute_error(y_true, y_pred)
    rmse = float(np.sqrt(mean_squared_error(y_true, y_pred)))
    r2   = r2_score(y_true, y_pred) if len(np.unique(y_true)) > 1 else np.nan
    mask = y_true != 0
    mape = float(np.mean(np.abs((y_true[mask]-y_pred[mask])/y_true[mask])) * 100) if mask.any() else np.nan
    acc  = max(0.0, 100.0 - mape) if not np.isnan(mape) else np.nan
    return dict(Accuracy=acc, MAE=mae, RMSE=rmse, R2=r2, ErrorRate=mape)


```

```python
# %% [9] UNIVARIATE FORECAST ========================================
def uni_holt(seri, h):
    s = pd.Series(seri).astype(float)
    if len(s) < 4: return np.repeat(s.iloc[-1], h)
    try:
        m = ExponentialSmoothing(s, trend="add", seasonal=None).fit(optimized=True)
        return np.clip(m.forecast(steps=h).values, 0, None)
    except: return np.repeat(s.iloc[-1], h)

def uni_arima(seri, h):
    s = np.asarray(seri, dtype=float)
    if len(s) < 6: return np.repeat(s[-1], h)
    best, ba = None, np.inf
    for o in [(1,1,1),(2,1,1),(0,1,1),(1,1,2),(2,1,2)]:
        try:
            m = ARIMA(s, order=o).fit()
            if m.aic < ba: best, ba = m, m.aic
        except: continue
    if best is None: return np.repeat(s[-1], h)
    return np.clip(np.asarray(best.forecast(steps=h)), 0, None)

def uni_prophet(yrs, vals, h):
    if not HAVE_PROPHET: return np.repeat(vals[-1], h)
    try:
        dp = pd.DataFrame({"ds": pd.to_datetime([f"{y}-01-01" for y in yrs]), "y": vals})
        m = Prophet(yearly_seasonality=False, daily_seasonality=False, weekly_seasonality=False)
        m.fit(dp)
        son = max(yrs)
        future = pd.DataFrame({"ds": pd.to_datetime([f"{son+i}-01-01" for i in range(1, h+1)])})
        return np.clip(m.predict(future)["yhat"].values, 0, None)
    except: return np.repeat(vals[-1], h)

def uni_forecast(ad, yrs, vals, h):
    if ad == "HoltWinters": return uni_holt(vals, h)
    if ad == "ARIMA":       return uni_arima(vals, h)
    if ad == "Prophet":     return uni_prophet(list(yrs), list(vals), h)
    raise ValueError(ad)


```

```python
# %% [10] EĞİTİM DÖNGÜSÜ ============================================
butun_sonuclar = []
butun_tahminler = []

cols = list(dict.fromkeys(["yıl"] + FEATURES + [HEDEF]))  # tekil
veri_h = veri[cols].dropna()
print(f"\nVeri (lag sonrası): {veri_h.shape}, yıl aralığı: "
      f"{veri_h['yıl'].min()}-{veri_h['yıl'].max()}")

# ML modelleri
for ad, fab in models_registry.items():
    for f_idx, (tr_y, te_y) in enumerate(folds, 1):
        tr = veri_h[veri_h["yıl"].isin(tr_y)]
        te = veri_h[veri_h["yıl"].isin(te_y)]
        if tr.empty or te.empty: continue
        X_tr, y_tr = tr[FEATURES].values, tr[HEDEF].values
        X_te, y_te = te[FEATURES].values, te[HEDEF].values
        try:
            m = fab()
            m.fit(X_tr, y_tr)
            pred = np.clip(m.predict(X_te), 0, None)
        except Exception as e:
            print(f"  [HATA] {ad}/fold{f_idx}: {e}")
            continue
        met = metrikler(y_te, pred)
        met.update(dict(Model=ad, Fold=f_idx,
                        TrainSon=max(tr_y), TestSon=max(te_y)))
        butun_sonuclar.append(met)
        if f_idx == 5:
            for yr, g, p in zip(te["yıl"].values, y_te, pred):
                butun_tahminler.append(dict(yıl=int(yr), model=ad,
                                            gercek=float(g), tahmin=float(p)))
    print(f"  {ad:<22} bitti")

# Univariate (HoltWinters, ARIMA, Prophet)
veri_uni = veri.dropna(subset=[HEDEF])
for ad in UNI_MODELS:
    for f_idx, (tr_y, te_y) in enumerate(folds, 1):
        tr = veri_uni[veri_uni["yıl"].isin(tr_y)]
        te = veri_uni[veri_uni["yıl"].isin(te_y)]
        if len(tr) < 6 or te.empty: continue
        pred = uni_forecast(ad, tr["yıl"].values, tr[HEDEF].values, len(te))
        y_te = te[HEDEF].values
        met = metrikler(y_te, pred)
        met.update(dict(Model=ad, Fold=f_idx,
                        TrainSon=max(tr_y), TestSon=max(te_y)))
        butun_sonuclar.append(met)
        if f_idx == 5:
            for yr, g, p in zip(te["yıl"].values, y_te, pred):
                butun_tahminler.append(dict(yıl=int(yr), model=ad,
                                            gercek=float(g), tahmin=float(p)))
    print(f"  {ad:<22} bitti (univariate)")


```

```python
# %% [11] METRİK ÖZETLER ============================================
sdf = pd.DataFrame(butun_sonuclar)
sdf.to_csv("butce_metrikleri_detay.csv", index=False, encoding="utf-8-sig")

genel = (sdf.groupby("Model")[["Accuracy","MAE","RMSE","R2","ErrorRate"]]
         .mean().round(3).reset_index()
         .sort_values("Accuracy", ascending=False))
genel.to_csv("butce_metrikleri_genel.csv", index=False, encoding="utf-8-sig")
print("\n========== GENEL MODEL KARŞILAŞTIRMASI ==========")
print(genel.to_string(index=False))

ana = sdf[sdf["Fold"] == 5]
en_iyi = ana.sort_values("Accuracy", ascending=False).head(1).iloc[0]
print(f"\n========== EN İYİ MODEL (Fold 5) ==========")
print(f"  Model:    {en_iyi['Model']}")
print(f"  Accuracy: {en_iyi['Accuracy']:.2f}%")
print(f"  MAE:      {en_iyi['MAE']:.2f}")
print(f"  R²:       {en_iyi['R2']:.4f}")
pd.DataFrame([en_iyi]).to_csv("butce_en_iyi_model.csv",
                              index=False, encoding="utf-8-sig")


```

```python
# %% [12] 2025-2030 RECURSIVE TAHMİN ================================
# Strateji:
#   * Model artık butce_TL değil, butce_USD tahmin eder.
#   * USD/TRY geleceği MODELLE TAHMİN EDİLMEZ.
#   * 2025-2030 USD/TRY değerleri doğrudan
#     usdtry_aylik_tahmin_2026_2030.csv dosyasından alınır.
#   * Aylık kur dosyası yıllık ortalama kura çevrilir.
#   * TL bütçe tahmini sonradan türetilir:
#       butce_TL_turetilen = butce_USD_tahmin × usdtry_hazir_veri

EN_IYI_MODEL = en_iyi["Model"]
H = 6
GELECEK_YILLAR = list(range(2025, 2031))
print(f"\nKullanılan model: {EN_IYI_MODEL}")
print(f"Model hedefi: {HEDEF}")

# ---------------------------------------------------------------------
# 1) 2025-2030 USD/TRY VERİSİNİ HAZIR DOSYADAN OKU
# ---------------------------------------------------------------------
GELECEK_KUR_PATHS = [
    "usdtry_aylik_tahmin_2026_2030.csv",
    "usdtry_aylik_tahmin_2025_2030.csv",
]

gelecek_kur_path = next((p for p in GELECEK_KUR_PATHS if os.path.exists(p)), None)
if gelecek_kur_path is None:
    raise FileNotFoundError(
        "2025-2030 USD/TRY tahmin dosyası bulunamadı. "
        "Beklenen dosya: usdtry_aylik_tahmin_2026_2030.csv"
    )

gelecek_kur_aylik = pd.read_csv(gelecek_kur_path)
gelecek_kur_aylik.columns = [str(c).strip().lower() for c in gelecek_kur_aylik.columns]

# Olası kolon adlarını güvenli yakala
yil_col = None
for aday in ["yıl", "yil", "year"]:
    if aday in gelecek_kur_aylik.columns:
        yil_col = aday
        break

if yil_col is None:
    # Tarih kolonu varsa yıldan üret
    tarih_adaylari = [c for c in gelecek_kur_aylik.columns if "tarih" in c or "date" in c]
    if tarih_adaylari:
        tarih_col = tarih_adaylari[0]
        gelecek_kur_aylik[tarih_col] = pd.to_datetime(gelecek_kur_aylik[tarih_col], errors="coerce")
        gelecek_kur_aylik["yıl"] = gelecek_kur_aylik[tarih_col].dt.year
        yil_col = "yıl"
    else:
        raise KeyError("Gelecek kur dosyasında yıl/yil/year veya tarih/date sütunu bulunamadı.")

kur_col = None
for aday in ["usdtry", "usd_try", "usd/try", "kur", "tahmin", "forecast", "prediction"]:
    if aday in gelecek_kur_aylik.columns:
        kur_col = aday
        break

if kur_col is None:
    adaylar = [c for c in gelecek_kur_aylik.columns if "usd" in c or "kur" in c or "try" in c]
    if not adaylar:
        raise KeyError("Gelecek kur dosyasında usdtry / kur sütunu bulunamadı.")
    kur_col = adaylar[0]

gelecek_kur_aylik[yil_col] = pd.to_numeric(gelecek_kur_aylik[yil_col], errors="coerce")
gelecek_kur_aylik[kur_col] = pd.to_numeric(
    gelecek_kur_aylik[kur_col].astype(str)
    .str.replace(",", ".", regex=False)
    .str.replace(" ", "", regex=False),
    errors="coerce"
)

# Aylık değerlerden yıllık ortalama kur üret
gelecek_kur_yillik = (
    gelecek_kur_aylik
    .dropna(subset=[yil_col, kur_col])
    .groupby(yil_col, as_index=False)[kur_col]
    .mean()
    .rename(columns={yil_col: "yıl", kur_col: "usdtry"})
)
gelecek_kur_yillik["yıl"] = gelecek_kur_yillik["yıl"].astype(int)
gelecek_kur_yillik = gelecek_kur_yillik[gelecek_kur_yillik["yıl"].isin(GELECEK_YILLAR)]

eksik_kur_yillari = sorted(set(GELECEK_YILLAR) - set(gelecek_kur_yillik["yıl"]))
if eksik_kur_yillari:
    raise ValueError(f"Gelecek USD/TRY dosyasında eksik yıllar var: {eksik_kur_yillari}")

gelecek_usd_dict = dict(zip(gelecek_kur_yillik["yıl"], gelecek_kur_yillik["usdtry"]))

print(f"[OK] Gelecek USD/TRY dosyadan okundu: {gelecek_kur_path}")
print("Yıllık ortalama USD/TRY:")
print(gelecek_kur_yillik.sort_values("yıl").to_string(index=False))

# ---------------------------------------------------------------------
# 2) Diğer gelecek varsayımları
# ---------------------------------------------------------------------
dev_ort = veri["devlet_orani"].tail(3).mean()
oz_ort  = veri["ozel_orani"].tail(3).mean()
print(f"Devlet oranı varsayımı: {dev_ort:.2f}%   |   Özel oranı: {oz_ort:.2f}%")

# Kişi başına USD için daha stabil büyüme varsayımı
kb_usd_growth = (veri["kisi_basi_USD"].iloc[-1] / veri["kisi_basi_USD"].iloc[-5]) ** (1/4) - 1
print(f"Kişi başına USD varsayım büyüme oranı: {kb_usd_growth*100:.2f}%/yıl")

# ---------------------------------------------------------------------
# 3) Recursive USD bütçe tahmini
# ---------------------------------------------------------------------
def recursive_predict_usd():
    """
    2025-2030 için USD bazlı bütçe tahmini yapar.
    TL bütçe burada tahmin edilmez; daha sonra hazır kurla türetilir.
    """

    if EN_IYI_MODEL in UNI_MODELS:
        seri = veri.dropna(subset=[HEDEF])
        pred_usd = uni_forecast(EN_IYI_MODEL, seri["yıl"].values, seri[HEDEF].values, H)
        yil_arr = np.array(GELECEK_YILLAR)
        usdtry_arr = np.array([gelecek_usd_dict[y] for y in yil_arr])
        return np.array(pred_usd), yil_arr, usdtry_arr

    # ML modeli tüm 1999-2024 verisiyle yeniden eğitilir
    train_full = veri_h.copy()
    X_full, y_full = train_full[FEATURES].values, train_full[HEDEF].values

    mdl = models_registry[EN_IYI_MODEL]()
    mdl.fit(X_full, y_full)

    # Son gözlenen durum: 2024
    son = veri.iloc[-1].copy()

    butce_usd_lag1 = son["butce_USD"]
    butce_usd_lag2 = veri.iloc[-2]["butce_USD"]
    butce_usd_yoy  = (butce_usd_lag1 - butce_usd_lag2) / butce_usd_lag2

    kb_usd_lag1 = son["kisi_basi_USD"]
    kb_usd_lag2 = veri.iloc[-2]["kisi_basi_USD"]
    kb_usd_yoy  = (kb_usd_lag1 - kb_usd_lag2) / kb_usd_lag2

    dev_lag1 = son["devlet_orani"]
    oz_lag1  = son["ozel_orani"]

    usdtry_lag1 = son["usdtry"]
    usdtry_lag2 = veri.iloc[-2]["usdtry"]
    usdtry_yoy  = (usdtry_lag1 - usdtry_lag2) / usdtry_lag2

    pred_usd_list = []
    yil_listesi = []
    usdtry_listesi = []

    for yr in GELECEK_YILLAR:
        usdtry_now = float(gelecek_usd_dict[yr])

        X_row_dict = {
            "yıl": yr,
            "butce_USD_lag1": butce_usd_lag1,
            "butce_USD_lag2": butce_usd_lag2,
            "butce_USD_yoy": butce_usd_yoy,
            "kisi_basi_USD_lag1": kb_usd_lag1,
            "kisi_basi_USD_lag2": kb_usd_lag2,
            "kisi_basi_USD_yoy": kb_usd_yoy,
            "devlet_orani_lag1": dev_lag1,
            "ozel_orani_lag1": oz_lag1,
            "usdtry_lag1": usdtry_lag1,
            "usdtry_yoy": usdtry_yoy,
        }

        X_arr = np.array([[X_row_dict[f] for f in FEATURES]])
        p_usd = float(np.clip(mdl.predict(X_arr), 0, None)[0])

        pred_usd_list.append(p_usd)
        yil_listesi.append(yr)
        usdtry_listesi.append(usdtry_now)

        # Bir sonraki yıl için lag güncelle — USD hedef üzerinden
        butce_usd_yoy = (p_usd - butce_usd_lag1) / butce_usd_lag1 if butce_usd_lag1 != 0 else 0
        butce_usd_lag2 = butce_usd_lag1
        butce_usd_lag1 = p_usd

        kb_usd_next = kb_usd_lag1 * (1 + kb_usd_growth)
        kb_usd_yoy = (kb_usd_next - kb_usd_lag1) / kb_usd_lag1 if kb_usd_lag1 != 0 else 0
        kb_usd_lag2 = kb_usd_lag1
        kb_usd_lag1 = kb_usd_next

        usdtry_yoy = (usdtry_now - usdtry_lag1) / usdtry_lag1 if usdtry_lag1 != 0 else 0
        usdtry_lag2 = usdtry_lag1
        usdtry_lag1 = usdtry_now

        dev_lag1 = dev_ort
        oz_lag1 = oz_ort

    return np.array(pred_usd_list), np.array(yil_listesi), np.array(usdtry_listesi)

pred_usd, yil_listesi, usd_listesi = recursive_predict_usd()
pred_tl = pred_usd * usd_listesi

```

```python
# %% [13] ÇIKTI =====================================================
sonuc = pd.DataFrame({
    "yıl": yil_listesi,
    "butce_USD_tahmin (Milyon USD)": np.round(pred_usd, 2),
    "usdtry_hazir_veri": np.round(usd_listesi, 4),
    "butce_TL_turetilen (Milyon TL)": np.round(pred_tl, 2),
})

print("\n========== 2025-2030 BÜTÇE TAHMİNİ — USD HEDEF, TL TÜRETİLMİŞ ==========")
print(sonuc.to_string(index=False))

# Gerçeklik kontrolü için büyüme oranları
sonuc["usd_yoy_%"] = sonuc["butce_USD_tahmin (Milyon USD)"].pct_change() * 100
sonuc["tl_yoy_%"] = sonuc["butce_TL_turetilen (Milyon TL)"].pct_change() * 100

print("\nYıllık büyüme kontrolü:")
print(sonuc[["yıl", "usd_yoy_%", "tl_yoy_%"]].round(2).to_string(index=False))

# Tarihsel + tahmin birleşik
tarihsel = veri[["yıl", "butce_TL", "usdtry", "butce_USD"]].copy()
tarihsel.columns = ["yıl", "butce_TL", "usdtry", "butce_USD"]
tarihsel["tip"] = "Gerçek"

tahmin = pd.DataFrame({
    "yıl": yil_listesi,
    "butce_TL": pred_tl,
    "usdtry": usd_listesi,
    "butce_USD": pred_usd,
    "tip": "Tahmin"
})

birlesik = pd.concat([tarihsel, tahmin], ignore_index=True)

sonuc.to_csv("butce_2025_2030_tahmin_usd_hedef.csv", index=False, encoding="utf-8-sig")
sonuc.to_excel("butce_2025_2030_tahmin_usd_hedef.xlsx", index=False)
birlesik.to_csv("butce_tarihsel_tahmin_birlesik_usd_hedef.csv", index=False, encoding="utf-8-sig")
birlesik.to_excel("butce_tarihsel_tahmin_birlesik_usd_hedef.xlsx", index=False)

print("\n[OK] Çıktılar:")
print("  - butce_2025_2030_tahmin_usd_hedef.xlsx")
print("  - butce_2025_2030_tahmin_usd_hedef.csv")
print("  - butce_tarihsel_tahmin_birlesik_usd_hedef.xlsx")
print("  - butce_tarihsel_tahmin_birlesik_usd_hedef.csv")
print("  - butce_metrikleri_genel.csv")
print("  - butce_en_iyi_model.csv")

```
